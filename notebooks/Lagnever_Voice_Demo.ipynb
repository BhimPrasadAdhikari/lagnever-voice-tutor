# Lagnever voice tutor: real-time demo

Student speaks -> **Vocal Bridge** (speech-to-text, turn-taking, bridge line) -> **tutor brain** (speech + notebook marks + student memory) -> **sympy** re-checks each step -> the page redraws while the tutor speaks.

Works in the DeepLearning.AI *Voice for AI Agents* lab or any Jupyter with the keys in `.env`.

```python
import os, sys, importlib
sys.path.insert(0, os.path.abspath('..'))
from dotenv import load_dotenv; load_dotenv(os.path.abspath('../.env'))
import checker, tutor, widget
for m in (checker, tutor, widget): importlib.reload(m)
from IPython.display import HTML
server_url = tutor.start_server(port=8766)
tutor.new_session(profile=tutor.JEE_STUDENT)   # class 12, JEE Main, two past slips on record
agent_id = os.environ['VOCAL_BRIDGE_AGENT_ID_LAGNEVER']
print('Tutor backend up:', server_url, '| model:', tutor.MODEL)
```
